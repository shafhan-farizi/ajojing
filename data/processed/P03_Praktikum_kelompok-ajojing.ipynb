```python
# TODO E2 — HASH
# Tulis fungsi samarkan(nilai, garam) yang mengembalikan 12 karakter pertama
# dari sha256(garam + nilai). Kembalikan np.nan bila nilainya kosong.
import hashlib

GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam):
  if pd.isna(nilai):
    return np.nan

  teks_gabungan = str(garam) + str(nilai)
  teks_bytes = teks_gabungan.encode('utf-8')

  hasil_sha256 = hashlib.sha256(teks_bytes).hexdigest()
  return hasil_sha256[:12]
# Terapkan pada kolom yang Anda beri tindakan 'hash'.
kolom_dihash = kamus[kamus['tindakan_penanganan'] == 'hash']['nama_kolom'].tolist()

for col in kolom_dihash:
  if col in aman.columns:
    aman[col] = aman[col].apply(lambda x: samarkan(x, GARAM))
# Expected: nilai asli dan hasil hash berbeda, dan nilai asli yang sama menghasilkan hash yang sama
for col in kolom_dihash:
  if col in aman.columns:
    print(f"Hasil hash kolom '{col}': {aman[col].dropna().iloc[0] if not aman[col].dropna().empty else "Data Kosong"}")

```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
# TODO F1 — tulis fungsi cek_k(data, kolom)
# Kembalikan Series berisi: k (grup terkecil), grup berisi 1 orang, total grup
def cek_k(data, kolom):
  ukuran_group = data.groupby(kolom, observed=True).size()

  k = ukuran_group.min()
  group_1_orang = (ukuran_group == 1).sum()
  total_group = len(ukuran_group)

  return pd.Series([k, group_1_orang, total_group], ['k', 'group_1_orang', 'total_group'])
```

```python
# TODO F2 — ukur risiko SEBELUM penanganan
# Gunakan data asli df dengan kolom:
#   ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
risiko_sebelum = cek_k(df, ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir'])
print(risiko_sebelum)
# Expected: k = 1 dan SELURUH 620 baris berada di grup berisi satu orang
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

Keempat kolom itu termasuk dalam kategori quasi-identifier, artinya jika nilai dari masing-masing kolom diketahui, maka semua data tersebut akan merujuk kepada kesimpulan 1 orang yang spesifik. Ditambah kemungkinan bahwa nilai k berarti 1, hal ini berarti kombinasi nilai 4 kolom akan merujuk pada 1 orang yang jelas.

```python
# TODO F3 — ukur risiko SESUDAH generalisasi
# Gunakan 'aman' dengan kolom:
#   ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
risiko_sesudah = cek_k(aman, ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur'])
print(risiko_sesudah)
# Expected: jumlah grup berisi satu orang turun banyak, tetapi k masih 1
```

```python
# TODO F4 — TANTANGAN: capai k >= 5
# Generalisasikan lebih jauh sampai k >= 5, dengan MEMPERTAHANKAN sebanyak mungkin informasi.
# Anda boleh: memperlebar kelompok umur, menggabungkan kecamatan kecil menjadi 'Lainnya',
# atau mengeluarkan satu kolom dari daftar quasi-identifier.
# Catat setiap pilihan dan dampaknya terhadap k.
# Petunjuk: ada lebih dari satu jawaban benar. Yang dinilai adalah alasannya.
aman_tantangan = aman.copy()

kecamatan_gabung = ['LIMO', 'BOJONGSARI', 'CIPAYUNG', 'SAWANGAN']
aman_tantangan['kecamatan'] = aman_tantangan['kecamatan'].apply(lambda x: 'Lainnya' if x in kecamatan_gabung else x)

cek_resiko = cek_k(aman_tantangan, ['jenis_kelamin', 'kecamatan'])
print(cek_resiko)

```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.

Kombinasi yang dipilih hanya kolom jenis kelamin dan kecamatan yang sudah sedikit dimodifikasi dengan menggabungkan beberapa kecamatan asal peserta yang jumlahnya kurang dari 45 sebagai "Lainnya". Dari kombinasi kedua kolom ini menghasilkan nilai k = 20, pemilihan ini dilakukan karena tingkat keunikan individu atau outlier sangatlah ekstrem. Maka, demi mencapai k>5, tindakan ini dilakukan, sehingga data yang akan dipublikasi akan aman dari risiko pelacakan identitas.

```python
# Membuat kamus data
kamus_team = pd.DataFrame({
    'nama_kolom': df_team.columns,
    'tipe_data': [str(df_team[col].dtype) for col in df_team.columns],
    'nilai_unik': [df_team[col].nunique() for col in df_team.columns],
    'nilai_kosong': [df_team[col].isnull().sum() for col in df_team.columns],
    'contoh_nilai': [df_team[col].dropna().iloc[0] if not df_team[col].dropna().empty else np.nan for col in df_team.columns]
})

# Membuat object untuk pemetaan kategori data per kolom
penandaan_team = {
    'Invoice ID': ('pengenal langsung', 'hash'),

    'Branch': ('quasi-identifier', 'pertahankan'),
    'City': ('quasi-identifier', 'pertahankan'),
    'Customer type': ('quasi-identifier', 'pertahankan'),
    'Gender': ('quasi-identifier', 'pertahankan'),
    'Date': ('quasi-identifier', 'pertahankan'),
    'Time': ('quasi-identifier', 'pertahankan'),

    'Product line': ('bukan data pribadi', 'pertahankan'),
    'Unit price': ('bukan data pribadi', 'pertahankan'),
    'Quantity': ('bukan data pribadi', 'pertahankan'),
    'Tax 5%': ('bukan data pribadi', 'pertahankan'),
    'Sales': ('bukan data pribadi', 'pertahankan'),
    'Payment': ('bukan data pribadi', 'pertahankan'),
    'cogs': ('bukan data pribadi', 'pertahankan'),
    'gross margin percentage': ('bukan data pribadi', 'pertahankan'),
    'gross income': ('bukan data pribadi', 'pertahankan'),
    'Rating': ('bukan data pribadi', 'pertahankan'),
}

list_tanda = []
for kolom, (kategori, tindakan) in penandaan_team.items():
  list_tanda.append({
      'nama_kolom': kolom,
      'kategori': kategori,
      'tindakan': tindakan
  })

tanda_team = pd.DataFrame(list_tanda)

# Memetakan setiap kategori data pada DataFrame kamus data
kamus_team['tindakan_penanganan'] = kamus_team['nama_kolom'].map(
    tanda_team.set_index('nama_kolom')['tindakan']
).fillna('pertahankan')
```
