**Data:** `kunjungan_perpustakaan_2025.csv` — catatan kunjungan perpustakaan kampus, Februari–Juni 2025.
Satu baris = satu kunjungan.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('kunjungan_perpustakaan_2025.csv')
df.head()
```

---
### Tugas 1 (15 poin) — Memuat dan memeriksa

```python
# a. Cetak jumlah baris dan kolom
print("Jumlah baris dan kolom:", df.shape)

# b. Cetak jumlah nilai kosong (missing) untuk setiap kolom
print("\nMissing value setiap kolom:")
print(df.isna().sum())

# c. Cetak berapa mahasiswa unik (nim_hash) yang pernah berkunjung
print("\nJumlah mahasiswa unik:", df["nim_hash"].nunique())
```

---
### Tugas 2 (20 poin) — Filter

```python
# Ambil kunjungan mahasiswa Teknik Informatika dengan durasi > 120 menit
ti = df[(df["prodi"] == "Teknik Informatika") & (df["durasi_menit"] > 120)]

# Cetak hasil
print("Jumlah baris:", len(ti))
print("Rata-rata jumlah_pinjam:", round(ti["jumlah_pinjam"].mean(), 2))
```

---
### Tugas 3 (20 poin) — Kolom turunan

```python
# a. Buat kolom 'durasi_jam' = durasi_menit / 60, dibulatkan 2 desimal
df["durasi_jam"] = (df["durasi_menit"] / 60).round(2)

# b. Buat kolom 'lama'
df["lama"] = np.where(df["durasi_menit"] >= 90, "Panjang", "Singkat")

# c. Tampilkan 5 kunjungan dengan durasi terlama
top5 = df.sort_values(by="durasi_menit", ascending=False)[
    ["prodi", "jenis_ruang", "durasi_jam"]
].head(5)

print(top5)
```

---
### Tugas 4 (25 poin) — Agregasi

```python
# Buat tabel ringkasan per 'jenis_ruang'
ringkas = (
    df.groupby("jenis_ruang")
      .agg(
          jumlah_kunjungan=("jenis_ruang", "count"),
          rata_durasi_menit=("durasi_menit", lambda x: round(x.mean(), 1)),
          total_jumlah_pinjam=("jumlah_pinjam", "sum"),
          rata_kepuasan=("kepuasan", lambda x: round(x.mean(), 2))
      )
      .sort_values(by="rata_durasi_menit", ascending=False)
)

# Tampilkan hasil
print(ringkas)
```

---
### Tugas 5 (10 poin) — Menyimpan hasil

```python
# Simpan tabel 'ringkas' ke file 'ringkasan_ruang.csv'
ringkas.to_csv("ringkasan_ruang.csv", index=True)

print("File 'ringkasan_ruang.csv' berhasil disimpan.")
```
